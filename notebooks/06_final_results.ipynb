```python
import pandas as pd
from IPython.display import display, Markdown

# ============================================================
# 1. PROJECT TITLE
# ============================================================

display(Markdown("""
# 🛒 Online Shoppers Purchase Prediction
## Final Model Results & Comparison

**Objective:** Predict whether an online shopping session will result in a purchase.
"""))


# ============================================================
# 2. INITIAL MODEL VALIDATION
# ============================================================

initial_results = pd.DataFrame({
    "Model": [
        "Random Forest",
        "Gradient Boosting",
        "Logistic Regression",
        "Decision Tree"
    ],
    "Training Accuracy": [0.9597, 0.9210, 0.8853, 0.9302],
    "Validation Accuracy": [0.8996, 0.8935, 0.8771, 0.8848],
    "Train-Val Difference": [0.0600, 0.0275, 0.0082, 0.0454],
    "Status": [
        "Good Generalization",
        "Good Generalization",
        "Good Generalization",
        "Good Generalization"
    ]
})

# Convert to percentage for display
initial_display = initial_results.copy()

initial_display["Training Accuracy"] = (
    initial_display["Training Accuracy"] * 100
).round(2).astype(str) + "%"

initial_display["Validation Accuracy"] = (
    initial_display["Validation Accuracy"] * 100
).round(2).astype(str) + "%"

initial_display["Train-Val Difference"] = (
    initial_display["Train-Val Difference"] * 100
).round(2).astype(str) + "%"

display(Markdown("## 1️⃣ Initial Model Validation"))
display(initial_display)


# ============================================================
# 3. CROSS VALIDATION
# ============================================================

cv_results = pd.DataFrame({
    "Model": [
        "Gradient Boosting",
        "Random Forest",
        "Logistic Regression",
        "Decision Tree"
    ],
    "Fold 1": [0.927237, 0.922094, 0.893143, 0.876688],
    "Fold 2": [0.921660, 0.920717, 0.862655, 0.872425],
    "Fold 3": [0.927731, 0.928442, 0.898319, 0.889063],
    "Fold 4": [0.939288, 0.930425, 0.896192, 0.901808],
    "Fold 5": [0.930647, 0.921248, 0.898193, 0.877191],
    "Mean ROC-AUC": [0.929312, 0.924585, 0.889701, 0.883435],
    "Std": [0.005773, 0.004032, 0.013652, 0.010724]
})

cv_display = cv_results.copy()

for col in [
    "Fold 1", "Fold 2", "Fold 3",
    "Fold 4", "Fold 5",
    "Mean ROC-AUC", "Std"
]:
    cv_display[col] = cv_display[col].round(4)

display(Markdown("## 2️⃣ Cross-Validation Results"))
display(cv_display)


# ============================================================
# 4. HYPERPARAMETER TUNING
# ============================================================

tuning_results = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest"
    ],
    "Tuning Method": [
        "GridSearchCV",
        "RandomizedSearchCV"
    ],
    "Best CV ROC-AUC": [
        0.914204,
        0.925033
    ]
})

tuning_results["Best CV ROC-AUC"] = (
    tuning_results["Best CV ROC-AUC"].round(4)
)

display(Markdown("## 3️⃣ Hyperparameter Tuning"))
display(tuning_results)


# ============================================================
# 5. FINAL TEST COMPARISON
# ============================================================

final_results = pd.DataFrame({
    "Rank": [1, 2, 3, 4, 5, 6],

    "Model": [
        "Gradient Boosting",
        "Random Forest Tuned",
        "Random Forest",
        "Decision Tree Tuned",
        "Logistic Regression",
        "Decision Tree"
    ],

    "Accuracy": [
        0.906186,
        0.905776,
        0.904957,
        0.902089,
        0.888570,
        0.896354
    ],

    "Precision": [
        0.720461,
        0.758503,
        0.756849,
        0.758123,
        0.754630,
        0.708738
    ],

    "Recall": [
        0.654450,
        0.583770,
        0.578534,
        0.549738,
        0.426702,
        0.573298
    ],

    "F1 Score": [
        0.685871,
        0.659763,
        0.655786,
        0.637329,
        0.545151,
        0.633864
    ],

    "ROC-AUC": [
        0.935737,
        0.928488,
        0.927721,
        0.920083,
        0.902461,
        0.880333
    ]
})

final_display = final_results.copy()

for col in ["Accuracy", "Precision", "Recall", "F1 Score"]:
    final_display[col] = (
        final_display[col] * 100
    ).round(2).astype(str) + "%"

final_display["ROC-AUC"] = final_display["ROC-AUC"].round(4)

display(Markdown("## 4️⃣ Final Test Set Comparison"))
display(final_display)


# ============================================================
# 6. BEST MODEL
# ============================================================

best_model = final_results.iloc[0]

display(Markdown(f"""
# 🏆 Final Best Model: {best_model["Model"]}

| Metric | Score |
|---|---:|
| Accuracy | **{best_model["Accuracy"]:.2%}** |
| Precision | **{best_model["Precision"]:.2%}** |
| Recall | **{best_model["Recall"]:.2%}** |
| F1 Score | **{best_model["F1 Score"]:.2%}** |
| ROC-AUC | **{best_model["ROC-AUC"]:.4f}** |
"""))


# ============================================================
# 7. FINAL CONCLUSION
# ============================================================

display(Markdown("""
## 📌 Final Conclusion

The **Gradient Boosting** model achieved the best overall performance.

### Key Results

- **Accuracy:** 90.62%
- **Precision:** 72.05%
- **Recall:** 65.45%
- **F1 Score:** 68.59%
- **ROC-AUC:** 0.9357
- **Cross-Validation Mean ROC-AUC:** 0.9293

Gradient Boosting was selected as the **final model** because it provided
the strongest overall performance across the important evaluation metrics.

---

### Project Flow

**EDA & Preprocessing**
→ **Feature Engineering**
→ **Model Training**
→ **Cross-Validation**
→ **Hyperparameter Tuning**
→ **Final Test Evaluation**
→ **Best Model: Gradient Boosting**
"""))
```
